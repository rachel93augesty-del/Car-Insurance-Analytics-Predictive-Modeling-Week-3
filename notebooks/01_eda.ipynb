```python
# =====================================
# Step 0 – Setup
# =====================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 20)
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (12,6)

# =====================================
# Step 1 – Load Data
# =====================================
raw_file = "../data/raw/raw_data.txt"  # Update path if needed

# Load with proper separator
df = pd.read_csv(raw_file, sep="|", low_memory=False)

print("Data Loaded Successfully")
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())

# =====================================
# Step 2 – Initial Inspection
# =====================================
# Preview first rows
display(df.head())

# Check data types
print("\nData Types:")
print(df.dtypes)

# Check missing values
print("\nMissing Values per Column:")
print(df.isna().sum())

# =====================================
# Step 3 – Data Cleaning
# =====================================
# Convert date columns
df['TransactionMonth'] = pd.to_datetime(df['TransactionMonth'], errors='coerce')
df['VehicleIntroDate'] = pd.to_datetime(df['VehicleIntroDate'], errors='coerce')

# Convert boolean columns
bool_cols = ['IsVATRegistered', 'AlarmImmobiliser', 'TrackingDevice']
for col in bool_cols:
    df[col] = df[col].astype(bool)

# Convert numeric columns with possible mixed types
numeric_cols = ['mmcode', 'Cylinders', 'cubiccapacity', 'kilowatts', 
                'NumberOfDoors', 'CustomValueEstimate', 'CapitalOutstanding',
                'SumInsured', 'CalculatedPremiumPerTerm', 'TotalPremium', 'TotalClaims']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Fill missing dates with median date
df['VehicleIntroDate'].fillna(df['VehicleIntroDate'].median(), inplace=True)

# Fill numeric missing values with median
for col in numeric_cols:
    df[col].fillna(df[col].median(), inplace=True)

# Fill object/categorical missing values with 'Unknown'
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    df[col].fillna('Unknown', inplace=True)

# Drop unnecessary column
if 'NumberOfVehiclesInFleet' in df.columns:
    df.drop(columns=['NumberOfVehiclesInFleet'], inplace=True)

print("\nData Types after Cleaning:")
print(df.dtypes)

print("\nMissing Values after Cleaning:")
print(df.isna().sum())

# =====================================
# Step 4 – Descriptive Statistics
# =====================================
print("\nNumeric Summary:")
display(df.describe())

print("\nCategorical Summary:")
for col in cat_cols:
    print(f"\n{col} Value Counts:")
    print(df[col].value_counts().head(5))  # Top 5

# =====================================
# Step 5 – Univariate Analysis
# =====================================
# Histograms for numeric variables
df[numeric_cols].hist(bins=30, figsize=(18,12))
plt.tight_layout()
plt.show()

# Bar plots for top categorical features
top_cat_features = ['Province', 'VehicleType', 'Gender', 'CoverCategory']
for col in top_cat_features:
    sns.countplot(y=col, data=df, order=df[col].value_counts().index[:10])
    plt.title(f"Top {col} counts")
    plt.show()

# =====================================
# Step 6 – Bivariate/Multivariate Analysis
# =====================================
# Correlation heatmap
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()

# Scatter plot: TotalPremium vs TotalClaims
sns.scatterplot(x='TotalPremium', y='TotalClaims', data=df, alpha=0.3)
plt.title("TotalPremium vs TotalClaims")
plt.show()

# Claims by Province
province_claims = df.groupby('Province')['TotalClaims'].sum().sort_values(ascending=False)
province_claims.plot(kind='bar')
plt.title("Total Claims by Province")
plt.ylabel("Total Claims")
plt.show()

# =====================================
# Step 7 – Outlier Detection
# =====================================
for col in ['TotalClaims', 'CustomValueEstimate']:
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()

# =====================================
# Step 8 – KPIs
# =====================================
# Loss ratio
df['LossRatio'] = df['TotalClaims'] / df['TotalPremium']
df['LossRatio'].replace([np.inf, -np.inf], np.nan, inplace=True)
df['LossRatio'].fillna(0, inplace=True)

# Summary by Province
province_summary = df.groupby('Province').agg(
    TotalPremium=('TotalPremium', 'sum'),
    TotalClaims=('TotalClaims', 'sum'),
    LossRatio=('LossRatio', 'mean')
).sort_values(by='LossRatio', ascending=False)
display(province_summary.head(10))

# =====================================
# Step 9 – Visualization of KPIs
# =====================================
# Loss ratio by VehicleType
vehicle_loss = df.groupby('VehicleType')['LossRatio'].mean().sort_values(ascending=False)
vehicle_loss.plot(kind='bar', color='orange')
plt.title("Average Loss Ratio by VehicleType")
plt.ylabel("Loss Ratio")
plt.show()
# =====================================
# Step 10 – Save Cleaned Data
# =====================================
processed_file = "../data/processed/insurance_data_cleaned.csv"

# Save to CSV
df.to_csv(processed_file, index=False)
print(f"Cleaned data saved to: {processed_file}")

```
