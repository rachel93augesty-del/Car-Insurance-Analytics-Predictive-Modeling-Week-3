```python
# ==========================
# Task 3: Hypothesis Testing Visualization
# ==========================

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ----------------------------
# Create images folder if not exists
# ----------------------------
os.makedirs("../images", exist_ok=True)

# ----------------------------
# Load cleaned data
# ----------------------------
data_path = "../data/processed/cleaned_data.csv"
if not os.path.exists(data_path):
    raise FileNotFoundError(f"Cleaned CSV not found at {data_path}")

df = pd.read_csv(data_path, low_memory=False)

# ----------------------------
# Fix column names if needed
# ----------------------------
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

# ==========================
# H1: Claim Frequency by Province
# ==========================
prov_freq = df.groupby("province")["claim_occurred"].mean().sort_values(ascending=False)
plt.figure(figsize=(10,6))
ax = sns.barplot(x=prov_freq.index, y=prov_freq.values, palette="Blues_d")
plt.ylabel("Claim Frequency")
plt.title("H1: Claim Frequency by Province")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../images/H1_Claim_Frequency_Province.png")
plt.show()

# ==========================
# H2: Claim Frequency by Zip Code (Top 10 for readability)
# ==========================
zip_freq = df.groupby("postalcode")["claim_occurred"].mean().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,6))
ax = sns.barplot(x=zip_freq.index.astype(str), y=zip_freq.values, palette="Greens_d")
plt.ylabel("Claim Frequency")
plt.title("H2: Claim Frequency by Zip Code (Top 10)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../images/H2_Claim_Frequency_ZipCode.png")
plt.show()

# ==========================
# H3: Margin by Zip Code (Top 10)
# ==========================
zip_margin = df.groupby("postalcode")["margin"].mean().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,6))
ax = sns.barplot(x=zip_margin.index.astype(str), y=zip_margin.values, palette="Oranges_d")
plt.ylabel("Average Margin")
plt.title("H3: Average Margin by Zip Code (Top 10)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../images/H3_Margin_ZipCode.png")
plt.show()

# ==========================
# H4: Claim Frequency by Gender
# ==========================
gender_freq = df.groupby("gender")["claim_occurred"].mean().sort_values(ascending=False)
plt.figure(figsize=(6,6))
ax = sns.barplot(x=gender_freq.index, y=gender_freq.values, palette="Purples_d")
plt.ylabel("Claim Frequency")
plt.title("H4: Claim Frequency by Gender")
plt.tight_layout()
plt.savefig("../images/H4_Claim_Frequency_Gender.png")
plt.show()

# ==========================
# H5: Claim Severity by Province
# ==========================
prov_severity = df.groupby("province")["claim_severity"].mean().sort_values(ascending=False)
plt.figure(figsize=(10,6))
ax = sns.barplot(x=prov_severity.index, y=prov_severity.values, palette="Reds_d")
plt.ylabel("Average Claim Severity")
plt.title("H5: Claim Severity by Province")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../images/H5_Claim_Severity_Province.png")
plt.show()
```
